```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="white")
df = pd.read_csv('technova_empleados.csv')
```

**Misión 1: La Distribución del Equipo (Comparación)**

La junta directiva quiere visualizar cómo se reparte el volumen total de empleados entre los distintos departamentos de la empresa para evaluar si están sobredimensionados en alguna área.

**Paso 1: Análisis del problema**

 **Tipo de variables**: Categórica (nominal) — departamento.

**Objetivo analítico:**
Comparar la cantidad de empleados entre departamentos para identificar si alguno está sobredimensionado.

**Decisión visual:**
Se utilizará un Gráfico de barras ya que permite comparar magnitudes entre categorías de forma clara;
se descarta el gráfico de torta porque con varios departamentos las proporciones se distinguen peor.





**Paso 2: Construcción de la visualización**

1. Agrupar por departamento

```python
# Paso 1: agrupar por departamento
grupos = df.groupby("departamento")
```

2. Obtener la cantidad de empleados

```python
# Paso 2: contar empleados en cada grupo
conteo_departamento = grupos["id_empleado"].count()
```

3.  Seleccionar el gráfico adecuado

```python
#Grafico de barras

fig, ax = plt.subplots(figsize=(12, 6))

ax.bar(conteo_departamento.index, conteo_departamento.values, color="steelblue")

ax.set_xlabel("Departamentos")
ax.set_title("Cantidad de empleados por departamento")
ax.set_ylabel("Cantidad de empleados")

plt.show()
```

4. Aplicar principios de limpieza visual

```python
sns.despine(ax=ax)
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

**Paso 3: Interpretación (Markdown)**

¿Qué departamento concentra la mayor cantidad de empleados?

El departamento que concentra mayor cantidad de empleados es el de Tecnología.

¿Existe alguna desproporción evidente en la estructura organizacional?

Sí existe una desproporcion ya que Tecnología y Ventas concentran más del 50% del equipo, mientras que Legal (10) y Operaciones (12) son las áreas más chicas.

 **Misión 2: El Histórico de Contrataciones (Tendencia)**

 Existen rumores de que la empresa sufrió una «congelación de contrataciones» en el pasado reciente debido a una crisis financiera. El Director quiere ver la evolución histórica año tras año para confirmarlo.

**Paso 1: Análisis del problema**

**Tipo de variables:**
anio_contratacion es una variable temporal (cuantitativa discreta) — representa una secuencia de años.

**Objetivo analítico:** Observar la evolución histórica (tendencia) de las contrataciones año a año, para detectar caídas que confirmen o descarten la supuesta "congelación de contrataciones".

**Decisión visual:**
Un gráfico de líneas (line chart). Es la elección correcta para mostrar tendencias a lo largo del tiempo — permite ver subidas, bajadas y quiebres en la serie de forma clara, algo que un gráfico de barras no comunica tan bien cuando hay muchos períodos.


**Paso 2: Construcción de la visualización**






1. Agrupar los datos por anio_contratacion

```python
# Agrupar por año de contratación
grupos_anio = df.groupby("anio_contratacion")
```

2. Contar la cantidad de contrataciones por año




```python
# Contar contrataciones por año
conteo_anio = grupos_anio["id_empleado"].count()
```

3. Construir el gráfico adecuado para series temporales

```python
#Grafico de Lineas(Line chart)
fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(conteo_anio.index, conteo_anio.values, marker="o", color="steelblue")

ax.set_title("Evolución de contrataciones por año")
ax.set_xlabel("Año de contratación")
ax.set_ylabel("Cantidad de contrataciones")

plt.tight_layout()
plt.show()
```

4. Aplicar buenas prácticas de diseño

```python
# Buenas prácticas de diseño
sns.despine(ax=ax)
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

**Paso 3: Interpretación (Markdown)**

¿En qué año se observa la caída más significativa de contrataciones?

En el año 2020 se observa la caída más significativa en contrataciones, con solo 2.

¿Se evidencia una recuperación posterior?

Si en 2021 y 2022 se evidencia una recuperación posterior

**Misión 3: Equidad Salarial (Correlación)**

Para evitar demandas laborales, la empresa necesita auditar su política salarial. Existe la hipótesis de que TechNova premia económicamente la edad del empleado. ¿Existe una relación matemática real entre la edad del trabajador y su salario en dólares?

**Paso 1: Análisis del problema**

**Tipo de variables:** edad y salario_usd, ambas cuantitativas continuas.

**Objetivo analítico:** Ver si existe correlación entre edad y salario.

**Decisión visual:** Se elegirá un Gráfico de dispersión (scatter plot), ideal para relación entre dos variables numéricas.

**Paso 2: Construcción de la visualización**

.Utilizar directamente las variables edad y salario_usd

.Seleccionar el gráfico adecuado para analizar correlaciones

.Construir una visualización clara y limpia


```python
#Grafico de dispersion(scatter plot)

fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(df["edad"], df["salario_usd"], color="steelblue", alpha=0.6)

ax.set_title("Relación entre edad y salario")
ax.set_xlabel("Edad")
ax.set_ylabel("Salario (USD)")

# Buenas prácticas de diseño
sns.despine(ax=ax)
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

 **1) Qué estás analizando?**

 Analicé si la edad influye en el salario de los empleados de TechNova.

**2) Hallazgo: Qué patrón observas (correlación positiva, negativa o nula)**

Hay una correlación positiva fuerte (0.84): a mayor edad, mayor salario.


**3) Impacto: Qué implicancia tiene esto para la política salarial de la empresa**

Esto confirma la hipótesis y podría ser un riesgo legal si el salario se basa en la edad y no en criterios objetivos como desempeño o puesto.
